```python
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import LogLocator, NullFormatter

plt.rcParams.update({
    "font.family":      "serif",
    "font.serif":       ["DejaVu Serif", "Times New Roman", "Times"],
    "font.size":        12,
    "axes.linewidth":   1.1,
    "mathtext.fontset": "dejavuserif",
    "savefig.dpi":      300,
})

# ---- energy window (all 5 sources present 3-20 MeV; we display 1-20) --------
E_MIN, E_MAX = 1.0, 20.0     # MeV

B_TO_MB   = 1000.0           # barns -> millibarns
EV_TO_MEV = 1e-6             # eV    -> MeV

# ---- ROBUST file finder: matches by keyword, ignores spaces/underscores -----
#  This is the fix for the "file not found" warnings: it does NOT depend on
#  exact file names. It looks for a signature word inside each filename.
def _norm(s):
    return s.lower().replace(" ", "").replace("_", "").replace("-", "")

def resolve(key):
    sigs = {"EXFOR": "exfor", "TENDL": "tendl", "ENDF": "endf",
            "JEFF": "jeff", "TALYS": "talys"}
    want = sigs[key]
    for fn in os.listdir("."):
        if fn.startswith("."):
            continue
        n = _norm(fn)
        # the JENDL/ENDF point file may be named with 'ebdf', 'endf', or 'jendl'
        if want in n or (key == "ENDF" and ("jendl" in n or "ebdf" in n)):
            return fn
    return None
```

```python
def parse_exfor(path):
    df = pd.read_csv(path)
    out = pd.DataFrame({
        "energy_MeV": pd.to_numeric(df["energy_MeV"], errors="coerce"),
        "xs_mb":      pd.to_numeric(df["cross_section_b"], errors="coerce") * B_TO_MB,
        "dE_MeV":     pd.to_numeric(df.get("d_energy_MeV", 0), errors="coerce").fillna(0),
        "dxs_mb":     pd.to_numeric(df.get("d_cross_section_b", 0), errors="coerce").fillna(0) * B_TO_MB,
        "author":     df.get("author", "Unknown").astype(str).str.strip(),
        "year":       pd.to_numeric(df.get("year", 0), errors="coerce").fillna(0).astype(int),
    })
    out = out.dropna(subset=["energy_MeV", "xs_mb"])
    out = out[(out["energy_MeV"] >= E_MIN) & (out["energy_MeV"] <= E_MAX)]
    out = out[out["xs_mb"] > 0]
    out["label"] = out["author"] + " (" + out["year"].astype(str) + ")"
    return out.reset_index(drop=True)

def parse_tendl(path):
    """TENDL/YANDF export. Header lines start with '#' or '##' and may contain
    'E,xs'. Data are already MeV / mb."""
    rows = []
    with open(path, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            s = line.strip().strip(',').strip()
            if not s or s.startswith("#") or s.startswith('"'):
                continue
            if "e,xs" in s.lower() or s.lower().startswith("e,"):
                continue
            parts = [p for p in s.replace(";", ",").split(",") if p != ""]
            if len(parts) < 2:
                continue
            try:
                rows.append((float(parts[0]), float(parts[1])))
            except ValueError:
                continue
    df = pd.DataFrame(rows, columns=["energy_MeV", "xs_mb"])
    df = df[(df["energy_MeV"] >= E_MIN) & (df["energy_MeV"] <= E_MAX)]
    df = df[df["xs_mb"] > 0].sort_values("energy_MeV")
    return df.reset_index(drop=True)

def parse_endf(path):
    """JENDL/ENDF point file: 'E[eV]  Sig[b]  Interp'. eV->MeV, b->mb."""
    rows = []
    with open(path, "r", encoding="utf-8", errors="ignore") as f:
        for line in f:
            s = line.strip()
            if not s or s.startswith("#"):
                continue
            parts = s.split()
            if len(parts) < 2:
                continue
            try:
                rows.append((float(parts[0]) * EV_TO_MEV, float(parts[1]) * B_TO_MB))
            except ValueError:
                continue
    df = pd.DataFrame(rows, columns=["energy_MeV", "xs_mb"])
    df = df[(df["energy_MeV"] >= E_MIN) & (df["energy_MeV"] <= E_MAX)]
    df = df[df["xs_mb"] > 0].sort_values("energy_MeV")
    return df.reset_index(drop=True)

def parse_jeff(path):
    df = pd.read_csv(path)
    if "Energy_MeV" in df.columns:
        e = pd.to_numeric(df["Energy_MeV"], errors="coerce")
    else:
        e = pd.to_numeric(df["Energy_eV"], errors="coerce") * EV_TO_MEV
    xs = pd.to_numeric(df["CrossSection_barns"], errors="coerce") * B_TO_MB
    out = pd.DataFrame({"energy_MeV": e, "xs_mb": xs}).dropna()
    out = out[(out["energy_MeV"] >= E_MIN) & (out["energy_MeV"] <= E_MAX)]
    out = out[out["xs_mb"] > 0].sort_values("energy_MeV")
    return out.reset_index(drop=True)

def parse_talys(path):
    """TALYS xlsx: ONE (E[MeV], sigma[mb]) block per LEVEL-DENSITY MODEL.
    Row 0 holds the model titles (block starts wherever a title string is
    found), row 1 the column headers, data from row 2. Sub-threshold rows
    contain '-' placeholders -> coerced to NaN and dropped.

    Returns a dict  {model_label: DataFrame(energy_MeV, xs_mb)}  — e.g.
    {'CTM': df, 'BFM/BSFM': df, 'GSM': df, 'Microscopic (Skyrme)': df}.
    Blocks are detected automatically, so extra models in the sheet are
    picked up too."""
    raw = pd.read_excel(path, header=None)
    titles = raw.iloc[0]

    def short_label(title):
        # last inner-most "( ... )" group, e.g. '... Model (CTM))' -> 'CTM'
        m = re.findall(r"\(([^()]*)\)", str(title))
        lab = (m[-1] if m else str(title)).strip().rstrip(")")
        lab = re.sub(r"\s*/\s*", "/", lab)            # 'BFM / BSFM' -> 'BFM/BSFM'
        return {"Skyrme force": "Microscopic (Skyrme)"}.get(lab, lab)

    models = {}
    for col, t in titles.items():
        if not isinstance(t, str) or not t.strip():
            continue                                  # separator column -> skip
        e  = pd.to_numeric(raw.iloc[2:, col],     errors="coerce")
        xs = pd.to_numeric(raw.iloc[2:, col + 1], errors="coerce")
        df = pd.DataFrame({"energy_MeV": e.values, "xs_mb": xs.values}).dropna()
        df = df[(df["energy_MeV"] >= E_MIN) & (df["energy_MeV"] <= E_MAX)]
        df = df[df["xs_mb"] > 0].sort_values("energy_MeV")
        if len(df):
            models[short_label(t)] = df.reset_index(drop=True)
    return models

PARSERS = {"EXFOR": parse_exfor, "TENDL": parse_tendl, "ENDF": parse_endf,
           "JEFF": parse_jeff, "TALYS": parse_talys}

data = {}
for key, parser in PARSERS.items():
    path = resolve(key)
    if path is None:
        print(f"[WARNING] {key}: no matching file uploaded — skipping.")
        data[key] = None
        continue
    try:
        data[key] = parser(path)
        if key == "TALYS":     # TALYS is now a dict of per-model DataFrames
            desc = ", ".join(f"{m} ({len(d)} pts)" for m, d in data[key].items())
            print(f"[OK] {key:5s}: {len(data[key])} level-density models -> {desc}")
            print(f"          from '{path}'")
        else:
            print(f"[OK] {key:5s}: {len(data[key]):4d} points  from '{path}'")
    except Exception as e:
        print(f"[ERROR] {key}: failed to parse '{path}': {e}")
        data[key] = None
```

```python
import matplotlib.ticker as mticker

# ============================================================================
# CELL 4 — Create the publication-quality comparison plot
# ============================================================================
fig, ax = plt.subplots(figsize=(11, 10))   # tall figure -> room for legends

# ---- evaluated libraries: one plain curve each ------------------------------
CURVE_STYLE = {
    "TENDL": dict(color="#1f9e89", ls="--", lw=2.2, label="TENDL-2023"),
    "ENDF":  dict(color="#e8702a", ls="-.", lw=2.2, label="ENDF/B (JENDL)"),
    "JEFF":  dict(color="#c0392b", ls=":",  lw=2.6, label="JEFF"),
}
for key in ["TENDL", "ENDF", "JEFF"]:
    df = data.get(key)
    if df is not None and len(df):
        ax.plot(df["energy_MeV"], df["xs_mb"], **CURVE_STYLE[key], zorder=3)

# ---- TALYS: one line + marker per level-density model -----------------------
TALYS_MODEL_STYLE = [
    dict(color="#3a0f6e", marker="o"),    # CTM
    dict(color="#7b2cbf", marker="s"),    # BFM/BSFM
    dict(color="#b5179e", marker="^"),    # GSM
    dict(color="#4361ee", marker="D"),    # Microscopic (Skyrme)
    dict(color="#5aa9e6", marker="v"),    # spare
    dict(color="#2a9d8f", marker="P"),    # spare
]

talys_handles, talys_labels = [], []
talys = data.get("TALYS")

if talys:
    for i, (model, df) in enumerate(talys.items()):
        st = TALYS_MODEL_STYLE[i % len(TALYS_MODEL_STYLE)]

        h, = ax.plot(
            df["energy_MeV"],
            df["xs_mb"],
            color=st["color"],
            ls="-",
            lw=2.0,
            marker=st["marker"],
            markersize=6,
            markevery=6,
            mfc="white",
            mec=st["color"],
            mew=1.4,
            label="_talys_" + model,
            zorder=4,
        )

        talys_handles.append(h)
        talys_labels.append(model)

# ---- EXFOR experimental points (TOP 10 DATASETS ONLY) -----------------------
exfor = data.get("EXFOR")

if exfor is not None and len(exfor):

    TOP_N = 10

    counts = exfor["label"].value_counts()
    top_labels = list(counts.index[:TOP_N])

    # Keep ONLY the top TOP_N EXFOR datasets
    exfor = exfor[exfor["label"].isin(top_labels)].copy()

    cmap = plt.cm.tab20(np.linspace(0, 1, len(top_labels)))
    markers = ["o", "s", "^", "v", "D", "<", ">", "p", "*", "h", "X", "P"]

    for i, lab in enumerate(top_labels):

        sub = exfor[exfor["label"] == lab]

        ax.errorbar(
            sub["energy_MeV"],
            sub["xs_mb"],
            yerr=sub["dxs_mb"].where(sub["dxs_mb"] > 0, np.nan),
            xerr=sub["dE_MeV"].where(sub["dE_MeV"] > 0, np.nan),
            fmt=markers[i % len(markers)],
            color=cmap[i],
            ecolor=cmap[i],
            markersize=5,
            capsize=3,
            elinewidth=0.9,
            capthick=0.9,
            linestyle="none",
            alpha=0.85,
            label=lab,
            zorder=2,
        )

ax.set_xscale("linear")
ax.set_yscale("log")
ax.set_xlim(0, 20)
ax.set_ylim(1, 2000)

# Set x-axis ticks
ax.set_xticks([0, 5, 10, 15, 20])
# Set y-axis ticks
ax.set_yticks([1, 10, 100, 1000])
ax.yaxis.set_major_formatter(mticker.ScalarFormatter())

ax.set_xlabel("Neutron Energy (MeV)", fontsize=13)
ax.set_ylabel("Cross Section (mb)", fontsize=13)

ax.set_title(
    "Fe-56(n,p)Mn-56 Cross-Section Comparison:\n"
    "EXFOR vs TALYS (all level-density models) vs TENDL vs ENDF vs JEFF",
    fontsize=14,
    pad=12,
)

ax.grid(which="major", ls="-", lw=0.6, alpha=0.5)
ax.grid(which="minor", ls=":", lw=0.4, alpha=0.3)

# ---- LEGEND 1 (TOP): TALYS level-density models -----------------------------
if talys_handles:
    leg_models = ax.legend(
        handles=talys_handles,
        labels=talys_labels,
        loc="upper center",
        bbox_to_anchor=(0.5, 1.0),
        ncol=min(len(talys_handles), 4),
        fontsize=9.5,
        title="TALYS level-density models",
        title_fontsize=10,
        framealpha=0.95,
        edgecolor="0.6",
        columnspacing=1.2,
        handletextpad=0.5,
        borderaxespad=0.3,
    )
    ax.add_artist(leg_models)

# ---- LEGEND 2: Libraries + EXFOR authors -----------------------------------
h_main, _ = ax.get_legend_handles_labels()

if h_main:
    ax.legend(
        loc="upper center",
        bbox_to_anchor=(0.5, 0.875),
        ncol=4,
        fontsize=8.5,
        framealpha=0.95,
        edgecolor="0.6",
    )

fig.tight_layout()

OUT_PNG = "Fe56_np_combined_comparison_plot.png"
fig.savefig(OUT_PNG, dpi=300, bbox_inches="tight")

print(f"Saved plot -> {OUT_PNG}")
```

```python
def iter_sources():
    """Yield (source_name, dataframe). Expands the TALYS dict so every
    level-density model becomes its own source, e.g. 'TALYS-CTM'."""
    for key, df in data.items():
        if key == "TALYS":
            for model, mdf in (df or {}).items():
                yield f"TALYS-{model}", mdf
        else:
            yield key, df

frames = []
for src, df in iter_sources():
    if df is None or not len(df):
        continue
    tmp = df.copy()
    tmp["source"] = src
    if src == "EXFOR":
        tmp = tmp[["source", "energy_MeV", "xs_mb", "dE_MeV", "dxs_mb", "author", "year"]]
    else:
        tmp = tmp[["source", "energy_MeV", "xs_mb"]]
        tmp["dE_MeV"] = np.nan; tmp["dxs_mb"] = np.nan
        tmp["author"] = np.nan; tmp["year"] = np.nan
    frames.append(tmp)

# guard: clear message instead of a cryptic KeyError if nothing loaded
if not frames:
    raise RuntimeError(
        "No data loaded — every source was skipped. Check the [WARNING] lines "
        "from Cell 3 and confirm all 5 files are uploaded (Files panel)."
    )

combined = pd.concat(frames, ignore_index=True)
combined = combined.sort_values(["source", "energy_MeV"]).reset_index(drop=True)
OUT_CSV = "Fe56_np_parsed_data.csv"
combined.to_csv(OUT_CSV, index=False)
print(f"Saved data -> {OUT_CSV}  ({len(combined)} rows)")

OUT_TXT = "Fe56_np_summary_statistics.txt"
with open(OUT_TXT, "w") as f:
    f.write("Fe-56(n,p)Mn-56 Cross-Section Data — Summary Statistics\n")
    f.write("=" * 60 + "\n")
    f.write(f"Comparison energy window: {E_MIN}-{E_MAX} MeV\n\n")
    for src, df in iter_sources():
        f.write(f"[{src}]\n")
        if df is None or not len(df):
            f.write("  (no data / file missing)\n\n"); continue
        f.write(f"  points        : {len(df)}\n")
        f.write(f"  energy range  : {df['energy_MeV'].min():.4g} - {df['energy_MeV'].max():.4g} MeV\n")
        f.write(f"  xs min        : {df['xs_mb'].min():.4g} mb\n")
        f.write(f"  xs max        : {df['xs_mb'].max():.4g} mb\n")
        if src == "EXFOR":
            f.write(f"  unique authors: {df['author'].nunique()}\n")
            f.write(f"  year range    : {int(df['year'].min())}-{int(df['year'].max())}\n")
        f.write("\n")
print(f"Saved stats -> {OUT_TXT}")
print(open(OUT_TXT).read())

```
